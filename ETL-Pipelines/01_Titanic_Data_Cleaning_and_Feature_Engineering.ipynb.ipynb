```python
plt.scatter(new_x, new_y, c=prediction, cmap='coolwarm', edgecolors='black', marker='X', s=200)
plt.text(new_x - 2, new_y - 2, f"Pred: {prediction[0]}", fontsize=12, color='black')
plt.xlabel("Height (in cms)")
plt.ylabel("Weight (in kgs)")
plt.title("T-Shirt Size Classification with KNN")
plt.show()
```
